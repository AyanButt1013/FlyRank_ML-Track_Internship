## 3. Success metric

*One metric you can defend. What number means 'good'?*

To evaluate whether the model or analysis actually works and provides value over a baseline heuristic, we evaluate:

* **Precision@K (e.g., Precision@20 or Precision@50):** Of the top $K$ pages recommended for human review, what proportion genuinely required updating or experienced a real traffic decline? High Precision@K ensures content teams do not waste time inspecting false positives.
* **Average Precision (AP) / ROC-AUC:** Evaluates how effectively the system ranks true opportunities higher than non-opportunities across the entire dataset.
* **Actionable Reason Code Coverage:** Every top-ranked page must be accompanied by at least one transparent reason code (e.g., `stale_visible_page`, `low_ctr_visible_page`) so an editor can act on the recommendation immediately without guessing.
* **Outperforming the Baseline:** The ML ranking must systematically beat a static baseline rule (e.g., sorting purely by content age or traffic drop) on Precision@K and Average Precision in holdout validation tests.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

While simple `IF/ELSE` conditions (like `IF age > 180 days AND impressions > 500 THEN flag`) are easy to write, they fail in complex search environments for several key reasons:

1. **Non-Linear Interactions Between Features:**
   A page that is 180 days old with high search impressions might be performing fine if it has high engagement and strong domain authority. Conversely, a 60-day-old page with thin word count and declining CTR might be dropping rapidly. Fixed rules treat signals independently, whereas ML models capture non-linear feature interactions naturally.

2. **Weighted Probability vs. Rigid Cutoffs:**
   Conditional statements rely on hard, arbitrary thresholds (e.g., position > 10, word count < 1,000). A page sitting just on the edge of a threshold gets missed entirely. ML models output a continuous probability score (0 to 1), allowing us to construct a smoothly ranked queue rather than a binary, unprioritized bucket.

3. **Handling High-Dimensionality and Volume Adjustments:**
   Search metrics vary across position tiers, search intents, and content types. Writing manual conditional rules for every combination leads to fragile, unmaintainable "spaghetti SQL." ML algorithms learn these nuanced relationships automatically while adjusting for search volume and noise.
